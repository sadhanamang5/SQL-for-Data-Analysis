```python
import sqlite3
import pandas as pd
from IPython.display import display

conn = sqlite3.connect("ecommerce_analysis.db")

def run_query(query):
    result = pd.read_sql_query(query, conn)
    display(result)
    return result

print("Function created successfully!")
```

```python
run_query("SELECT * FROM Customers;")
```

```python
query1 = """
WITH ProductSales AS (
    SELECT
        c.region,
        p.product_name,
        SUM(oi.quantity * oi.unit_price) AS revenue
    FROM Customers c
    JOIN Orders o ON c.customer_id = o.customer_id
    JOIN Order_Items oi ON o.order_id = oi.order_id
    JOIN Products p ON oi.product_id = p.product_id
    WHERE o.status = 'Delivered'
    GROUP BY c.region, p.product_name
),
RankedProducts AS (
    SELECT *,
        RANK() OVER (
            PARTITION BY region
            ORDER BY revenue DESC
        ) AS product_rank
    FROM ProductSales
)
SELECT *
FROM RankedProducts
WHERE product_rank <= 3
ORDER BY region, product_rank;
"""

result1 = run_query(query1)
```

```python
query2 = """
WITH MonthlySales AS (
    SELECT
        strftime('%Y-%m', o.order_date) AS month,
        SUM(oi.quantity * oi.unit_price) AS revenue
    FROM Orders o
    JOIN Order_Items oi ON o.order_id = oi.order_id
    WHERE o.status = 'Delivered'
    GROUP BY strftime('%Y-%m', o.order_date)
),
Growth AS (
    SELECT *,
        LAG(revenue) OVER (ORDER BY month) AS previous_revenue
    FROM MonthlySales
)
SELECT
    month,
    revenue,
    previous_revenue,
    ROUND(
        (revenue - previous_revenue) * 100.0 /
        NULLIF(previous_revenue, 0), 2
    ) AS growth_percentage
FROM Growth;
"""

result2 = run_query(query2)
```

```python
query3 = """
WITH LatestDate AS (
    SELECT MAX(order_date) AS max_date
    FROM Orders
),
CustomerActivity AS (
    SELECT
        c.customer_id,
        c.customer_name,
        MAX(o.order_date) AS last_purchase
    FROM Customers c
    LEFT JOIN Orders o
        ON c.customer_id = o.customer_id
        AND o.status = 'Delivered'
    GROUP BY c.customer_id, c.customer_name
)
SELECT
    ca.customer_id,
    ca.customer_name,
    ca.last_purchase
FROM CustomerActivity ca
CROSS JOIN LatestDate ld
WHERE ca.last_purchase IS NULL
   OR ca.last_purchase < date(ld.max_date, '-90 days');
"""

result3 = run_query(query3)
```

```python
query4 = """
SELECT
    p.category,
    SUM(oi.quantity * oi.unit_price) AS total_revenue
FROM Products p
JOIN Order_Items oi ON p.product_id = oi.product_id
JOIN Orders o ON oi.order_id = o.order_id
WHERE o.status = 'Delivered'
GROUP BY p.category
ORDER BY total_revenue DESC;
"""

result4 = run_query(query4)
```

```python
query5 = """
WITH CustomerSpending AS (
    SELECT
        c.region,
        c.customer_name,
        SUM(oi.quantity * oi.unit_price) AS total_spending
    FROM Customers c
    JOIN Orders o
        ON c.customer_id = o.customer_id
    JOIN Order_Items oi
        ON o.order_id = oi.order_id
    WHERE o.status = 'Delivered'
    GROUP BY c.region, c.customer_id, c.customer_name
),
RankedCustomers AS (
    SELECT *,
        ROW_NUMBER() OVER (
            PARTITION BY region
            ORDER BY total_spending DESC
        ) AS rn
    FROM CustomerSpending
)
SELECT
    region,
    customer_name,
    total_spending
FROM RankedCustomers
WHERE rn = 1;
"""

result5 = run_query(query5)
```
